# ECHO: Raw Yelp Dataset Exploration

This notebook is the first exploratory pass over the raw Yelp Open Dataset. It checks the shape, fields, representative values, and a few distributions before the data is moved into MySQL during Phase 1.

The large source files are explored with samples here. Complete profiling results are recorded in [`docs/data_profile.txt`](../docs/data_profile.txt), while Phase 1 cleaning observations are maintained in [`docs/data_notes.md`](../docs/data_notes.md).

**Goals of this exploration**

- Understand what each source table contains.
- Identify nested, missing, unusually long, or inconsistent fields.
- Check which fields will need parsing or transformation during ETL.
- Record observations without modifying the raw files.



```python
import pandas as pd
from pathlib import Path
```

```python
RAW = Path("..") / "data" / "raw"
```

## 1. Business table

The business table contains one row per business. This section checks its schema, nested fields, categories, location coverage, ratings, and review-count distribution.



```python
def raw_path(name):
    return next(RAW.rglob(f"yelp_academic_dataset_{name}.json"))
```

```python
business = pd.read_json(raw_path("business"), lines=True, dtype=False, convert_dates=False)
business.head()
```

```python
business.info()
```

```python
business["attributes"].iloc[0]
```

```python
business['categories'].dropna().str.split(",").explode().value_counts().head(20)
```

```python
business["city"].value_counts().head(10)
```

```python
business["stars"].value_counts().sort_index()
```

```python
business["review_count"].value_counts().sort_index()
```

```python
business["state"]
```

## 2. Review table

The review table is the largest source file, so this notebook loads a 50,000-row sample rather than the full dataset. The full-file row count and field profile are available in `docs/data_profile.txt`.

## 3. Check-in table

The check-in source is stored at business level: each row contains a business and a comma-separated string of timestamps. This format will need to be exploded into one row per check-in during ETL.



```python
checkin = pd.read_json(raw_path("checkin"), lines=True, nrows=5, dtype=False, convert_dates=False)
```
